```python
import torch
import numpy as np
```

```python
tens_manual = torch.tensor([[1, 2, 3], [4, 5, 6], [6, 7, 8], [9, 10, 11]])
print(tens_manual)
tens_zeros = torch.zeros((4, 3)) #4 rows, 3 columns
print(tens_zeros)
tens_ones = torch.ones((4, 3))
print(tens_ones)
tens_rand = torch.rand((4, 3)) #random values between 0 and 1
print(tens_rand)

#So far tensors are created as columns,rows.
```

```python
print(f"Shape of tens_rand: {tens_rand.shape}")
#Shape attribute shows the dimensions of the tensor - rows and columns
```

```python
#Can you make a tensor of non-number data types?

non_numeric_tensor = torch.tensor([["apple", "banana"], ["cherry", "date"]])
print(non_numeric_tensor)   

#This will raise an error because PyTorch tensors only support numeric data types.
```

```python
#Now try making a tensor from a numpy array
np_array = np.array([[1, 2, 3], [4, 5, 6]])
tensor_from_np = torch.from_numpy(np_array)
print(tensor_from_np)
```

```python
#Compare numpy to tensor operations
np_sum = np_array + 10
print(f"Numpy array after adding 10:\n{np_sum}") 
tensor_sum = tensor_from_np + 10
print(f"Tensor after adding 10:\n{tensor_sum}")

#Maths operations are element-wise in both numpy and tensors
```

```python
#Add tensors together
tens_add = tens_rand + tens_ones
print(f"tens_rand + tens_ones:\n{tens_add}")

#Element-wise addition of two tensors of the same shape
#If shapes differ, an error will occur

tens_multiply = tens_rand * tens_ones
print(f"tens_rand * tens_ones:\n{tens_multiply}")
#Element-wise multiplication of two tensors of the same shape
#If shapes differ, an error will occur
```

```python
#Matrix multiplication
#tens_matmul = torch.matmul(tens_rand, tens_rand)
#print(f"Matrix multiplication of tens_rand with itself:\n{tens_matmul}")
#Matrix multiplication requires multiple tensors
#The number of columns in the first tensor must equal the number of rows in the second tensor
tens_matmul2 = torch.matmul(tens_rand.T, tens_rand)
print(f"Matrix multiplication of transposed tens_rand with tens_rand:\n{tens_matmul2}")
```
