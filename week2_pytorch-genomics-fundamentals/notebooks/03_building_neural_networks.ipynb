```python
import torch
import torch.nn as nn

class MyModel(nn.Module):
    def __init__(self):
        super().__init__()  # Always call parent init
        
        # Define layers here
        self.layer1 = nn.Linear(10, 5)  # 10 inputs → 5 outputs
        self.layer2 = nn.Linear(5, 1)   # 5 inputs → 1 output
    
    def forward(self, x):
        # Define how data flows through layers
        x = self.layer1(x)
        x = self.layer2(x)
        return x

# Create instance
model = MyModel()
print(model)
```

```python
# Create a simple linear layer
layer = nn.Linear(in_features=10, out_features=5)
# 3 inputs → 2 outputs
# Parameters: weight (2x3) + bias (2) = 8 total parameters

# Check the parameters
print("Weight shape:", layer.weight.shape)  # (2, 3)
print("Bias shape:", layer.bias.shape)      # (2,)
print("Weight:\n", layer.weight)
print("Bias:", layer.bias)

# Use it
x = torch.tensor([1.0, 2.0, 3.0])
output = layer(x)
print("Output:", output)  # 2 values
```
