```python
import torch
import torch.nn.functional as F

def simple_attention(query, key, value):
    """
    Compute attention mechanism.
    
    Args:
        query: (batch, seq_len, d_model) - what we're looking for
        key:   (batch, seq_len, d_model) - what we have
        value: (batch, seq_len, d_model) - what we output
    
    Returns:
        output: (batch, seq_len, d_model) - attended representation
        attention_weights: (batch, seq_len, seq_len) - where we focused
    """
    # 1. Compute similarity: Q·K^T
    d_model = query.size(-1)
    scores = torch.matmul(query, key.transpose(-1, -2))  # (batch, seq_len, seq_len)
    # 2. Scale by √d
    scaled_scores = scores / torch.sqrt(torch.tensor(d_model, dtype=torch.float32))
    # 3. Apply softmax
    attention_weights = F.softmax(scaled_scores, dim=-1)  # (batch, seq_len, seq_len)
    # 4. Multiply by V
    output = torch.matmul(attention_weights, value)  # (batch, seq_len, d_model)
    # 5. Return output + weights (for visualization)
    return output, attention_weights
```

```python
# Test
batch, seq_len, d_model = 10, 10, 4
Q = torch.rand(batch, seq_len, d_model)
K = torch.rand(batch, seq_len, d_model)
V = torch.rand(batch, seq_len, d_model)

output, weights = simple_attention(Q, K, V)
print(f"Output shape: {output.shape}")  # (2, 10, 4)
print(f"Attention weights shape: {weights.shape}")  # (2, 10, 10)
print(f"Attention weights sum: {weights[0, 0].sum()}")  # Should be 1.0
```
